## String

```python
text = "Python Programming"

print(len(text))
print(text.upper())
print(text.lower())
print(text[3])
print(text[0:6])
print("Python" in text)
print(text.replace("P", "x"))
```

## List

```python
arr = [10, 20, 30, 40, 50]

arr.append(60)
print(arr)

arr.insert(2, 5)
print(arr)

arr.remove(10)
print(arr)

arr.pop()
print(arr)

print(arr.index(30))
```

```python
student = ["deep", 21, "Bsc", 85.5]

print(student)

student.append("India")
print(student)

student.insert(1, "Man")
print(student)

student.remove("deep")
print(student)

student.pop()
print(student)

print(student.index(85.5))
```
